```python
from src.utils import set_seed

seed = 1238
set_seed(seed)
```

```python
from src.utils import get_norm_parameters

yaml_path = '/content/drive/MyDrive/data'
mean, std = get_norm_parameters(yaml_path = yaml_path)
print(mean)
print(std)
```

```python
from torchvision import transforms

test_transf = transforms.Compose([ transforms.Resize(256),
                                  transforms.CenterCrop(224),
                                  transforms.ToTensor(),
                                  transforms.Normalize(mean,std)
                                ])
```

```python
from src.food101_dataset import Food101DataSet

data_root = '/content'
test_csv = '/content/drive/MyDrive/data/test.csv'
test_ds = Food101DataSet(data_root = data_root, csv = test_csv, transform = test_transf)
```

```python
from torch.utils.data import DataLoader

test_loader = DataLoader(test_ds, batch_size = 1024, num_workers = 2, shuffle = False)
```

```python
import torch
from src.bitenet_v1 import BiteNetV1

ckpt_file_path = '/content/drive/MyDrive/results/bitenetv1/model/exp_1_2-40.pth'
ckpt_file = torch.load(ckpt_file_path)

model = BiteNetV1()
model.load_state_dict(ckpt_file['model'])
```
